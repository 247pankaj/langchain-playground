### Models Integration with OpenAI, Google Gemini, and Groq
Practice: unified `init_chat_model`, direct providers, streaming, batch.

```python
# Load API keys
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]   = os.getenv("GROQ_API_KEY")
```

```python
# OpenAI(Unified interface)
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
model
```

```python
response = model.invoke("What is machine learning in simple terms?")
response.content
```

```python
# Direct ChatOpenAI
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-4.1")
response = model.invoke("What is machine learning in simple terms?")
response
```

### Google Gemini

```python
# Unified interface (Gemini)
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.8-flash")
model.invoke("Why is the sky blue?").content
```

```python
# Direct Google Gemini
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
model.invoke("Why is the sky blue?")
```

### Groq

```python
# Unified interface (Groq)
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3.8-27b")
model.invoke("How do fireflies produce light?").content
```

```python
# Direct Groq
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3.8-27b")
model.invoke("How do fireflies produce light?")
```

### Streaming
Calling `.stream()` yields chunks as they are produced — better UX for long answers.

```python
# Non-streaming
model.invoke("Explain the water cycle in about 150 words.")
```

```python
# Streaming
for chunk in model.stream("Explain the water cycle in about 150 words."):
    print(chunk.text, end="|", flush=True)
```

### Batch
Run several independent requests in parallel.

```python
responses = model.batch(
    [
        "Why do cats purr?",
        "How do magnets work?",
        "What causes earthquakes?",
    ],
    config={"max_concurrency": 5},
)
for r in responses:
    print(r)
    print("-" * 60)
```
